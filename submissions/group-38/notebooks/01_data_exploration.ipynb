## Load datasets

```python
import sys
from pathlib import Path
```

```python
# locate shared project folder
project_dir = Path.cwd().resolve()
if project_dir.name == "notebooks":
    project_dir = project_dir.parent

# allow importing data_utils.py from parent folder
sys.path.insert(0, str(project_dir))

from data_utils import load_data, split_data
```

```python
# dataset stored locally in project's ignored data folder
zip_path = project_dir / "data" / "cornell-tech-cs-5785-combined-2026-ai-vs-human-text.zip"

# load datasets using shared function
train, unlabeled, test = load_data(zip_path)

# check dimensions of each dataset
print("Training:", train.shape)
print("Unlabeled:", unlabeled.shape)
print("Test:", test.shape)

train.head()
```

## Train/validation split

```python
# 80/20 stratified train/validation split
# human (0), AI (1), random_state=42
train_df, val_df = split_data(train)

# check size and label distribution of both sets
print("Training examples:", len(train_df))
print("Validation examples:", len(val_df))

print("\nTraining labels:")
print(train_df["label"].value_counts())

print("\nValidation labels:")
print(val_df["label"].value_counts())

```

## Dataset exploration

```python
# check for missing vals and duplicate text passages
for name, df in [("train", train), ("unlabeled", unlabeled), ("test", test)]:
    print(f"\n{name} dataset:")
    print("missing values:", df.isnull().sum().sum())
    print("duplicate text passages:", df.duplicated(subset="text").sum())

# examine lengths of text passages in training set
word_ct = train["text"].str.split().str.len()

print("\ntraining passage lengths (words):")
print(word_ct.describe())
```

### Conclusion from data exploration
There are no missing values or duplicate passages within each dataset. The 10k passages examined range from 75 to 350 words, with 209 words being the average length and 206 words being the median length. 

These results indicate that no missing-value imputation or deduplication within datasets is needed before modeling. Having passage length and these additional metrics may be useful later on for analysis and determining whether model performance differs with text length.
