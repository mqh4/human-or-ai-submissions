# Load datasets

```python
import pandas as pd
import zipfile
from pathlib import Path
from sklearn.model_selection import train_test_split
```

```python
# path to datasets
zip_path = Path.home() / "Downloads" / "AML" / "cornell-tech-cs-5785-combined-2026-ai-vs-human-text.zip"

# load from ZIP
with zipfile.ZipFile(zip_path) as z:
    folder = "cornell-tech-cs-5785-combined-2026-ai-vs-human-text/"

    train = pd.read_csv(z.open(folder + "train.csv"))
    unlabeled = pd.read_csv(z.open(folder + "unlabeled.csv"))
    test = pd.read_csv(z.open(folder + "test.csv"))

# check dimensions of each dataset
print("Training:", train.shape)
print("Unlabeled:", unlabeled.shape)
print("Test:", test.shape)

# preview first couple rows of training dataset
train.head()
```

# Train/validation split

```python
# split labeled training dataset into training and validation sets
# 80% training, 20% validation, stratified by label
# human (0), AI (1)
# fix random_state for reproducibility
train_df, val_df = train_test_split(
    train,
    test_size=0.2,
    stratify=train["label"],
    random_state=42
)

# check size and label distribution of both sets
print("Training examples:", len(train_df))
print("Validation examples:", len(val_df))

print("\nTraining labels:")
print(train_df["label"].value_counts())

print("\nValidation labels:")
print(val_df["label"].value_counts())
```
